```python
import zipfile

ZIP_PATH = r"C:\Users\sudha\Downloads\Sentinel project data.zip"  # change if your zip is elsewhere
TABLES = "sentinel_data/tables/"
NUM_PARTS = 2  # 2 x 5,000 = 10,000 transactions

z = zipfile.ZipFile(ZIP_PATH)
all_files = z.namelist()

tx_files = [TABLES + f"transactions/train/part-{i:05d}.csv.gz" for i in range(1, NUM_PARTS + 1)]
label_files = [TABLES + f"labels/train/part-{i:05d}.csv.gz" for i in range(1, NUM_PARTS + 1)]
customer_files = [f for f in all_files if f.startswith(TABLES + "customers/all/") and f.endswith(".csv.gz")]
merchant_files = [f for f in all_files if f.startswith(TABLES + "merchants/all/") and f.endswith(".csv.gz")]
device_files = [f for f in all_files if f.startswith(TABLES + "devices/all/") and f.endswith(".csv.gz")]

tx_df = pd.concat([pd.read_csv(z.open(f), compression="gzip") for f in tx_files])
labels_df = pd.concat([pd.read_csv(z.open(f), compression="gzip") for f in label_files])
cust_df = pd.concat([pd.read_csv(z.open(f), compression="gzip") for f in customer_files])
merch_df = pd.concat([pd.read_csv(z.open(f), compression="gzip") for f in merchant_files])
dev_df = pd.concat([pd.read_csv(z.open(f), compression="gzip") for f in device_files])
```

```python
tx_df.head()
```

```python
cust_df.head()
```

```python

# join fraud label, merchant name and device kind onto each transaction
tx_df = tx_df.merge(labels_df[["transaction_id", "fraud_label"]], on="transaction_id", how="left")
tx_df = tx_df.merge(merch_df[["merchant_id", "synthetic_name"]], on="merchant_id", how="left")
tx_df = tx_df.merge(dev_df[["device_id", "device_kind"]], on="device_id", how="left")
tx_df["transaction_time"] = pd.to_datetime(tx_df["transaction_time"])

# keep only the customers who appear in these transactions
cust_df = cust_df[cust_df["customer_id"].isin(tx_df["customer_id"])]
print(len(tx_df), "transactions,", len(cust_df), "customers,", int(tx_df["fraud_label"].sum()), "fraud")

# transactions: customer, merchant and device fields
tx_docs = []
for row in tx_df.to_dict("records"):
    tx_docs.append({
        "_id": row["transaction_id"],
        "customer_id": row["customer_id"],
        "transaction_time": row["transaction_time"].to_pydatetime(),
        "amount": row["transaction_amount"],
        "payment_channel": row["payment_channel"],
        "merchant": {
            "merchant_id": row["merchant_id"],
            "merchant_name": row["synthetic_name"],
            "merchant_category": row["merchant_category"],
            "location": json.loads(row["merchant_location"]),
        },
        "device": {
            "device_id": None if pd.isna(row["device_id"]) else row["device_id"],
            "device_kind": None if pd.isna(row["device_kind"]) else row["device_kind"],
        },
        "fraud_label": int(row["fraud_label"]),
    })

# devices each customer used, taken from the transactions
devices = tx_df.dropna(subset=["device_id"])[["customer_id", "device_id", "device_kind"]].drop_duplicates()
devices_by_customer = {}
for row in devices.to_dict("records"):
    devices_by_customer.setdefault(row["customer_id"], []).append(
        {"device_id": row["device_id"], "device_kind": row["device_kind"]}
    )

# customers with their devices embedded
cust_docs = []
for row in cust_df.to_dict("records"):
    cust_docs.append({
        "_id": row["customer_id"],
        "customer_age": row["customer_age"],
        "home_city": row["home_city"],
        "region": row["region"],
        "devices": devices_by_customer.get(row["customer_id"], []),
    })

customers.insert_many(cust_docs)
transactions.insert_many(tx_docs)
print("customers:", customers.count_documents({}))
print("transactions:", transactions.count_documents({}))
```

```python
transactions.create_index([("customer_id", ASCENDING)])
transactions.create_index([("transaction_time", DESCENDING)])
print(list(transactions.index_information()))

# AFTER: IXSCAN on customer_id_1 (reads only that customer's documents)
plan = transactions.find(query).sort("transaction_time", DESCENDING).explain()
print("AFTER")
print("  docs examined:", plan["executionStats"]["totalDocsExamined"])
print("  rows returned:", plan["executionStats"]["nReturned"])
pprint(plan["queryPlanner"]["winningPlan"])
```

```python
# 6a. Fraud rate per merchant_category
pipeline = [
    {"$group": {
        "_id": "$merchant.merchant_category",
        "transactions": {"$sum": 1},
        "fraud": {"$sum": "$fraud_label"},
    }},
    {"$project": {
        "transactions": 1,
        "fraud": 1,
        "fraud_rate_pct": {"$round": [{"$multiply": [{"$divide": ["$fraud", "$transactions"]}, 100]}, 2]},
    }},
    {"$sort": {"fraud_rate_pct": -1}},
]
for doc in transactions.aggregate(pipeline):
    print(doc)
```

```python
# 6b. Top 10 customers by amount in their last 24 hours of activity
# For each customer: find their latest transaction time, keep transactions
# in the 24 hours before it, add up the amount.
pipeline = [
    {"$setWindowFields": {
        "partitionBy": "$customer_id",
        "output": {"last_activity": {"$max": "$transaction_time"}},
    }},
    {"$match": {"$expr": {"$gte": [
        "$transaction_time",
        {"$dateSubtract": {"startDate": "$last_activity", "unit": "hour", "amount": 24}},
    ]}}},
    {"$group": {
        "_id": "$customer_id",
        "amount_last_24h": {"$sum": "$amount"},
        "transactions_last_24h": {"$sum": 1},
    }},
    {"$sort": {"amount_last_24h": -1}},
    {"$limit": 10},
    {"$project": {"amount_last_24h": {"$round": ["$amount_last_24h", 2]}, "transactions_last_24h": 1}},
]
for doc in transactions.aggregate(pipeline):
    print(doc)
```
